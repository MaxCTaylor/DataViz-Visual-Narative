````python
import pandas as pd
import plotly.express as px

cars = pd.read_csv("data/auto_mpg.csv")

# Use these in every plot so your colors, legend order, and labels match the targets
ORIGIN_ORDER = ["USA", "Europe", "Japan"]
ORIGIN_COLORS = {"USA": "#2a78d6", "Europe": "#eb6834", "Japan": "#1baf7a"}
LABELS = {
    "mpg": "Fuel economy (mpg)",
    "weight": "Vehicle weight (lb)",
    "horsepower": "Horsepower",
    "displacement": "Displacement (cu in)",
    "model_year": "Model year",
    "origin": "Origin",
}

cars.head()
````

## Plot 1: Scatter plot with marginal plots

- **Target:** `target-plots/plot-1_python_scatter-marginals.html`
- **You will practice:** a basic scatter plot, color groups, hover names, marginal plots

**What to reproduce**

- `weight` on the x axis, `mpg` on the y axis, one color per origin
- A histogram of weight **above** the scatter plot and box plots of mpg on the **right**
- Hovering over a point shows the **car name** in bold

**Hints**

- Start with `fig1 = px.scatter(cars, x="weight", y="mpg", color="origin")`, then add the standard arguments from the tips above
- Marginal plots are one argument each: `marginal_x="histogram"`, `marginal_y="box"`
- The bold car name in the hover comes from `hover_name="name"`
- Show the figure with `fig1.show()`
- Docs: [Scatter plots](https://plotly.com/python/line-and-scatter/), [Marginal plots](https://plotly.com/python/marginal-plots/)

````python
# Plot 1: your code here

fig1 = px.scatter(cars, x="weight", y="mpg", 
                  marginal_x = "histogram",
                  marginal_y= "box",
                  hover_name= "name",
                  color="origin",
                  color_discrete_map=ORIGIN_COLORS,
                  category_orders={"origin": ORIGIN_ORDER},
                  labels=LABELS,
                  template="plotly_white",
                  )

fig1.show()
````

## Plot 2: Animated bubble chart

- **Target:** `target-plots/plot-2_python_animated-scatter.html`
- **You will practice:** bubble sizes, animation with a play button and slider

**What to reproduce**

- `horsepower` on the x axis, `mpg` on the y axis, colored by origin
- Bubble **size** set by `displacement`
- One animation frame per `model_year`, with a play button and slider
- Axes stay fixed while it plays: x from 40 to 240, y from 5 to 50

**Hints**

- New arguments for `px.scatter()`: `size="displacement"`, `animation_frame="model_year"`
- Fix the axes with `range_x=[40, 240]` and `range_y=[5, 50]`. Try leaving them out and pressing play to see why they matter
- Keep `hover_name="name"` from Plot 1

````python
# Plot 2: your code here

fig2 = px.scatter(cars, x = "horsepower", y = "mpg",
                 range_x=[40,240],
                 range_y=[5,50],
                 color = "origin",
                 size = "displacement",
                 hover_name = "name",
                 animation_frame= "model_year",
                 color_discrete_map=ORIGIN_COLORS,
                 category_orders={"origin": ORIGIN_ORDER},
                 labels=LABELS,
                 template="plotly_white",
                 title = "Fuel Economy vs. Horsepower by Model Year",
                 )

fig2.show()
````

## Plot 3: Violin plots

- **Target:** `target-plots/plot-3_python_violin.html`
- **You will practice:** distribution plots, box plots inside violins, showing every point

**What to reproduce**

- One violin of `mpg` for each origin, each a different color
- A small **box plot** inside each violin
- **All** the individual cars shown as points beside each violin, with the car name on hover

**Hints**

- Start with `fig3 = px.violin(cars, x="origin", y="mpg", color="origin")`
- `box=True` adds the box plots, and `points="all"` shows every car
- Docs: [Violin plots](https://plotly.com/python/violin/)

````python
# Plot 3: your code here

fig3 = px.violin(cars, x = "origin", y = "mpg",
                 color = "origin",
                 box = True,
                 points = "all",
                 hover_name = "name",
                 color_discrete_map=ORIGIN_COLORS,
                 category_orders={"origin": ORIGIN_ORDER},
                 labels=LABELS,
                 template="plotly_white",
                 title = "Fuel Economy by Region of Origin",
                 )


fig3.show()
````

## Plot 4: Line chart with a range slider

- **Target:** `target-plots/plot-4_python_line-rangeslider.html`
- **You will practice:** summarizing data first, line charts, range sliders

**What to reproduce**

- The **average** mpg for each model year and origin, one line per origin with markers
- A **range slider** under the x axis. Drag its ends to zoom in on a range of years

**Hints**

- First compute the averages. This line is given to you:

  ```python
  yearly = cars.groupby(["model_year", "origin"], as_index=False)["mpg"].mean()
  ```

- Then plot the new `yearly` data frame: `fig4 = px.line(yearly, x="model_year", y="mpg", color="origin", markers=True, ...)`
- Add the slider after creating the figure: `fig4.update_xaxes(rangeslider_visible=True)`
- Docs: [Line charts](https://plotly.com/python/line-charts/), [Range slider](https://plotly.com/python/range-slider/)

````python
# Plot 4: your code here

yearly = cars.groupby(["model_year", "origin"],
                      as_index = False)["mpg"].mean()

fig4 = px.line(yearly, x = "model_year", y = "mpg",
               color = "origin",
               markers = True,
               color_discrete_map=ORIGIN_COLORS,
               category_orders={"origin": ORIGIN_ORDER},
               labels=LABELS,
               template="plotly_white", 
               title = "Average Fuel Economy by Model Year",
               )

fig4.update_xaxes(rangeslider_visible=True)

fig4.show()
````

## Bonus A: dropdown menu with graph objects

*Optional, worth 1 bonus point. Skip it if you like; leave the cell empty or delete it.*

- **Target:** `target-plots/bonus-A_python_dropdown-menu.html`
- **You will practice:** `plotly.graph_objects`, building traces yourself, dropdown menus

**What to reproduce**

- The same yearly averages as Plot 4, but for `mpg`, `horsepower`, **and** `weight`
- Built with `plotly.graph_objects`, not Plotly Express: one `go.Scatter` trace per origin
- A **dropdown menu** at the top right that switches the lines between the three variables and updates the y axis title
- Hovering on a year shows all three origins in one box

**Hints**

- This one is a step up, and it is optional. Plotly Express builds traces for you; here you build them yourself.
- Averages: `cars.groupby(["model_year", "origin"], as_index=False)[["mpg", "horsepower", "weight"]].mean()`
- Build the figure: `import plotly.graph_objects as go`, `figA = go.Figure()`, then loop over `ORIGIN_ORDER` and call `figA.add_trace(go.Scatter(x=..., y=..., name=origin, mode="lines+markers", line=dict(color=ORIGIN_COLORS[origin])))`
- A dropdown is a list of buttons in `figA.update_layout(updatemenus=[dict(buttons=[...])])`. Each button replaces the y data of all three traces (in order) and changes the axis title:

  ```python
  dict(label="Horsepower", method="update",
       args=[{"y": [usa_values, europe_values, japan_values]},
             {"yaxis.title.text": "Horsepower"}])
  ```

- Combined hover: `hovermode="x unified"` in `update_layout`
- Docs: [Graph objects](https://plotly.com/python/graph-objects/), [Dropdown menus](https://plotly.com/python/dropdowns/)

````python
# Plot A: your code here

import plotly.graph_objects as go

Averages = cars.groupby(
    ["model_year", "origin"], 
    as_index=False
    )[["mpg", "horsepower", "weight"]].mean()

figA = go.Figure()

for origin in ORIGIN_ORDER:
    subset = Averages[Averages["origin"] == origin]

    figA.add_trace(
        go.Scatter(x = subset["model_year"], y = subset["mpg"],
                   name = origin,
                   mode = "lines+markers",
                   line = dict(color=ORIGIN_COLORS[origin]),
                   )
                   ) 

figA.update_layout(updatemenus=[
    dict(
        x=1,
        xanchor="right",
        y=1.15,
        yanchor="top",
        buttons=[
    dict(
         label = "Fuel economy (mpg)", method = "update",
         args=[
             {"y": [Averages[Averages["origin"] == origin]["mpg"]
              for origin in ORIGIN_ORDER
              ]},
             {"yaxis.title.text": "Fuel economy (mpg)"}
             ]
             ), 
     dict(
         label = "Horsepower", method = "update",
           args=[
               {"y": [Averages[Averages["origin"] == origin]["horsepower"]
              for origin in ORIGIN_ORDER
              ]},
               {"yaxis.title.text": "Horsepower"}
               ]
               ),
     dict(
            label = "Vehicle weight (lb)", method = "update",
            args=[
                {"y": [Averages[Averages["origin"] == origin]["weight"]
              for origin in ORIGIN_ORDER
              ]},
                {"yaxis.title.text": "Vehicle weight (lb)"}
                ]
            ),
            ]   
        )
    ],
    xaxis_title = "Model year",
    hovermode = "x unified",
    template = "plotly_white"
)

figA.show()
````
